### 1. Read File

- open file
- read file
- close file

```python
file = open('shopping_list.txt', 'r')
content = file.read()
file.close()
print(content)
```

As you can see that, the values are one below other insted of side by side because,

- when every we press enter in the txt file then it will automatically takes it as \n so when every it prints it will print every element in the new line

```python
# the Raw Representation of the .txt file
print(repr(content))
```

### 2. Write File

- open file
- write file
- close file

```python
items = "mango\napple\nsugar"
file = open('grocery.txt', 'w')
file.write(items)
file.close()
```

### 3. Append File

```python
file = open('grocery.txt', 'a') #a - append
items = '\ncurd'
file.write(items)
file.close()
```

### For avoiding to forget the file.close() we can use CONTEXT MANAGERS

```python
with open('shopping_list.txt', 'a') as file:
  file.write('\npizza')
```
